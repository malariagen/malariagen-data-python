# Diplotype clustering

Two related methods on `Ag3` for hierarchically clustering **diplotypes** (a sample's pair of haplotypes at each site, i.e. its diploid genotype) across a genome region, and rendering the result as an interactive dendrogram. If you use either in a publication, cite both this package and doi.org/10.1093/molbev/msae140.

### Setup

```python
import malariagen_data
ag3 = malariagen_data.Ag3(
    "simplecache::gs://vo_agam_release_master_us_central1",
    simplecache=dict(cache_storage="../../gcs_cache"),
    results_cache="../../results_cache",
)
ag3
```

## `plot_diplotype_clustering`

Clusters samples by genotype similarity over a region and plots the resulting dendrogram. Key parameters:

- **`region`**: genome region to cluster over (contig, `"contig:start-end"`, or a gene/transcript ID). Smaller regions run faster and give a more locus-specific tree; whole-chromosome regions are much slower.
- **`sample_sets`** / **`sample_query`**: which samples to include - either named sample sets/releases, or a pandas query string against sample metadata (e.g. `"taxon == 'arabiensis'"`) to select a specific subset.
- **`cohort_size`**: if set, randomly down-samples to this many samples (raises an error if fewer are available) - useful to keep large cohorts tractable.
- **`site_mask`**: which site-filter mask to apply before clustering (e.g. `"gamb_colu"`, `"gamb_colu_arab"`, or `None` for no masking) - changes which SNPs are considered reliable enough to use.
- **`color`** / **`symbol`**: a metadata column name (or a `{label: query}` mapping to define custom groups) used to colour/symbol-code the leaves - lets you visually check whether clusters correspond to known groupings like taxon or country.
- **`linkage_method`** (`'single'`, `'complete'`, `'average'`, `'weighted'`, `'centroid'`, `'median'`, `'ward'`) and **`distance_metric`** (`'cityblock'`, `'euclidean'`, `'sqeuclidean'`): control how the hierarchical clustering itself is computed - changing these can noticeably reshape the tree topology.
- **`count_sort`** / **`distance_sort`**: control the left-right ordering of dendrogram branches (by descendant count or by distance) - purely cosmetic, doesn't change the clustering itself.
- **`render_mode`** (`'svg'` vs `'webgl'`): rendering backend - `webgl` is faster for very large trees.

```python
ag3.plot_diplotype_clustering(
    region="2L:2,410,000-2,430,000",
    sample_sets=["AG1000G-GH", "AG1000G-BF-B"],
    site_mask="gamb_colu",
    color="taxon",
    symbol="country",
    linkage_method="complete",
    distance_metric="cityblock",
    width=1000,
    height=500,
    count_sort=True,
    distance_sort=False,
)
```

## `plot_diplotype_clustering_advanced`

Same underlying dendrogram as above, but with extra annotation tracks stacked beneath it: per-sample heterozygosity, amino-acid variant genotypes for chosen transcripts, and/or gene CNV state - so you can see how those relate to the clustering. Key additional parameters beyond the ones shared with `plot_diplotype_clustering` above:

- **`heterozygosity`** (bool) / **`heterozygosity_colorscale`**: whether to draw the heterozygosity track, and its colour scale.
- **`snp_transcript`**: one or more transcript IDs to plot amino-acid variant genotypes for (as a track under the dendrogram); `None` omits this track entirely.
- **`snp_filter_min_maf`**: drops amino-acid variants below this minor allele frequency, to declutter the SNP track.
- **`snp_query`**: a pandas query string restricting which SNP effect types are shown (defaults to non-synonymous/start-lost/stop-lost/stop-gained - i.e. amino-acid-changing variants).
- **`cnv_region`**: genome region to plot gene CNV calls for (as another track); often the same as `region`, but can differ - `None` omits the CNV track.
- **`cnv_max_coverage_variance`**: excludes samples whose coverage variance exceeds this threshold before computing CNV calls (removes noisy/unreliable coverage profiles).
- **`dendrogram_height`** / **`heterozygosity_height`** / **`snp_row_height`** / **`cnv_row_height`**: pixel heights for each track, for layout control.

```python
ag3.plot_diplotype_clustering_advanced(
    region="2R:28,480,000-28,500,000",
    cnv_region="2R:28,480,000-28,500,000",
    snp_transcript="AGAP002862-RA",
    snp_filter_min_maf=0.05,
    sample_sets="AG1000G-GH",
    site_mask="gamb_colu",
    color="taxon",
    symbol="country",
    linkage_method="complete",
    count_sort=True,
    distance_sort=False,
    show=False,
)
```
